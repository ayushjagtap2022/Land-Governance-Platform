## 10-Line Python SDK Demonstration Notebook
This notebook demonstrates programmatic policy search, 640-district GIS querying, and side-by-side policy simulation using `land_governance_sdk`.

```python
from land_governance_sdk import create_client

# 1. Initialize Client (Set offline=True for standalone notebook demo without running server)
client = create_client(offline=True)

# 2. Search Policy Guidelines
docs = client.documents.search('drone survey', state='Maharashtra')
print(f"Found {docs.count} guidelines: {docs.documents[0].title}")

# 3. Query GIS District Indicators & Convert to Pandas DataFrame
df_districts = client.gis.get_districts(state='Maharashtra').to_dataframe()
display(df_districts[['district', 'dispute_risk', 'modernization_index']].head())

# 4. Run Policy Simulations and Compare Scenarios Side-by-Side
scen_a = client.simulation.run(policy_variable='digital_cadastre', target_value=60.0, investment_cr=50.0)
scen_b = client.simulation.run(policy_variable='digital_cadastre', target_value=95.0, investment_cr=250.0)
cmp = client.simulation.compare(scen_a, scen_b)
print(f"Net Litigation Savings: ₹{cmp.deltas['litigation_savings_delta_cr']} Cr | {cmp.winner}")
```
